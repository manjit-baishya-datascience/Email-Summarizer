```python
import pandas as pd
import numpy as np
import torch
import matplotlib.pyplot as plt
from collections import Counter

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import evaluate

import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

# Data these libraries need at runtime
nltk.download("wordnet",   quiet=True)     # synonym lookups for METEOR
nltk.download("omw-1.4",   quiet=True)     # multilingual wordnet data METEOR relies on
nltk.download("punkt",     quiet=True)     # sentence/word tokenizer models
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)     # common words ("the", "is", ...) our extractive summarizer will ignore

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

## **Example Sentences**

- **Simple sentence**

```python
import textwrap

source_simple = (
    """The Amazon rainforest produces about 20% of the world's oxygen supply. Scientists warn that rapid deforestation is threatening the rainforest's ability to keep producing that oxygen."""
)

reference_simple = (
    "The Amazon, a major source of the world's oxygen, is being endangered by deforestation."
)

print("--- Simple Passage ---")
print(textwrap.fill(source_simple, width = 100))
print("\nReference summary:", textwrap.fill(reference_simple, width = 100))
```

- **Dialogue format sentence**

```python
example_dialogue = (
    "Mike: Have you met up with this girl Tom's been talking about?\n"
    "Tom: Not yet, but we're meeting tomorrow at the coffee shop.\n"
    "Mike: Nice! Let me know how it goes.\n"
    "Tom: Will do, I'm a bit nervous though.\n"
    "Mike: Just be yourself, you'll be fine."
)

reference_dialogue = (
    "Tom is meeting a girl at a coffee shop tomorrow and feels a bit nervous. "
    "Mike will find out how it went and reassures him."
)

print("--- Conversation ---")
print(textwrap.fill(example_dialogue, width = 100))
print("\nReference summary:", textwrap.fill(reference_dialogue, width = 100))
```
